Next we define two more functions **CleanDf** and **SplitDf** below. 

The function **CleanDf** cleans the given database by removing entries that don't have a result, and converting the string result values into floats off 1, 0.5, 0 for white win, draw and black win respectively. It also sets the type of each column in the dataframe to make manipulating data easier in the future. Finally **CleanDf** takes parameters of the dataframe to be cleaned and a rating gap, which eliminates all games that have a difference in Elo of the two players higher than the given rating gap. This is to ensure the games considered for analysis will be between players of relatively close ability.

The function **SplitDf** takes our dataframe and a rating increment parameter, and then creates new dataframes of games in a specific rating range, for instance if the rating increment is 100 (which is the current default) the game database will be split into dataframes: rating[0], rating[1200], rating[1300], rating[1400] ... rating[2800], rating[2900], rating[3000]. These dataframes can then be analysed to see how the openings vary at different ratings.

```python
#Game_db_new = pd.DataFrame()
#ImportAllGames(10)

def CleanDf(df,Rating_gap = 100): #Note this currently eliminates all games where one persons rating is 0, could change to be equal to others?
    global Game_db_new
    
    df = df[df["Result"].isin(["1-0","1/2-1/2","0-1"])] #Removes any games with different result
    df.loc[:,"Result"] = df["Result"].replace({"1-0":1, "1/2-1/2":0.5,"0-1":0}) #Converts result to numbers
    df = df.astype({"Opening":"string","BlackElo":int,"WhiteElo":int,"Result":float,"Moves":"string"}) #typecasts each column "WhiteElo":int
    df = df[abs(df["WhiteElo"] - df["BlackElo"]) < Rating_gap] #Shows only games within a certain rating
    

    Game_db_new = df
    return Game_db_new

def SplitDf(Rating_increment, df = pd.DataFrame()): #FIX WHAT TO DO WITH 0 RATING, also last rating increment isn't created.

    df_copy = df #creates copy to use
    global Rating_df #creates df to output
    Rating_df = {}
    i = 1200
    global Rating_cieling #creates list of ratings
    Rating_cieling = [0]
    while i in range(1200,3000): #makes a list of ratings from 1200 to 30000 incrementing by rating increment
        Rating_cieling.append(i)
        i = i + Rating_increment
    print(Rating_cieling)

    j = 0
    try:
        df_copy.insert(3, "AverageElo", (df["WhiteElo"] + df["BlackElo"])/2,allow_duplicates=False) #inserts new column "Average Elo"
    except Exception as error:
        print(f"Error is :{Exception}, and {error}")
    #print(df)
    while j < len(Rating_cieling)-1:

        Rating_df[Rating_cieling[j]] = df_copy[df_copy["AverageElo"].isin(range(Rating_cieling[j],Rating_cieling[j+1]))] #Stores entire dataframe in a dictionary
        j = j + 1
    #print(df)
    return
    
```

```python
def OpeningSummary(df, NoMoves): #Want Dict with Opening moves: OPening moves, winrate: winrate, number games: number games

    Dict={}
    Opening_moves = []
    Number_of_Games = []
    Number_of_Wins = []
    Win_rate = []
    Number_of_Losses = []
    Loss_rate = []
    Number_of_Draws = []
    Draw_rate = []
    Rating_difference = []
    Average_rating_difference = []
    Average_score = []
    #df = Df
    df = df.reset_index(drop = True,inplace = False)
    
    for i in range(0,df.shape[0]-1):
        moves = df.loc[i,"Moves"]
        moves = moves.replace(". ",".")#replace( "2.","").replace("3.","").replace( "4.","").replace( "5.","") #should be able to delete line actually
        move_list = " ".join(moves.split()[0:NoMoves]) #converts into string
        if move_list not in Opening_moves: #Gives list of all moves - would have to change
            Opening_moves.append(move_list)            
            Number_of_Games.append(1)
            Rating_difference.append(df.loc[i,"WhiteElo"]-df.loc[i,"BlackElo"])
            if df.loc[i,"Result"] == 1:
                Number_of_Wins.append(1)
                Number_of_Draws.append(0)
                Number_of_Losses.append(0)
            elif df.loc[i,"Result"] == 0.5:
                Number_of_Wins.append(0)
                Number_of_Draws.append(1)
                Number_of_Losses.append(0)
            else:
                Number_of_Wins.append(0)
                Number_of_Draws.append(0)
                Number_of_Losses.append(1)
        else:
            Number_of_Games[Opening_moves.index(move_list)] = Number_of_Games[Opening_moves.index(move_list)] + 1
            Rating_difference[Opening_moves.index(move_list)] = Rating_difference[Opening_moves.index(move_list)] + df.loc[i,"WhiteElo"]-df.loc[i,"BlackElo"]
            if df.loc[i,"Result"] == 1:
                Number_of_Wins[Opening_moves.index(move_list)] = Number_of_Wins[Opening_moves.index(move_list)] + 1
            elif df.loc[i,"Result"] == 0.5:
                Number_of_Draws[Opening_moves.index(move_list)] = Number_of_Draws[Opening_moves.index(move_list)] + 1
            else:
                Number_of_Losses[Opening_moves.index(move_list)] = Number_of_Losses[Opening_moves.index(move_list)] + 1
                
    for j in range(0, len(Number_of_Wins)):
        Win_rate.append(round(Number_of_Wins[j]/Number_of_Games[j],4)*100)
        Draw_rate.append(round(Number_of_Draws[j]/Number_of_Games[j],4)*100)
        Loss_rate.append(round(Number_of_Losses[j]/Number_of_Games[j],4)*100)
        Average_rating_difference.append(round(Rating_difference[j]/Number_of_Games[j],2))
        Average_score.append(round((Win_rate[j]+0.5*Draw_rate[j])/100,2))
        

    
    Dict = {"Opening":Opening_moves,"Win rate(%)":Win_rate,"Loss rate(%)":Loss_rate,"Draw rate(%)":Draw_rate,"Number of Games":Number_of_Games,"Average Score":Average_score,"Rating Difference":Average_rating_difference}
    dfSummary = pd.DataFrame(Dict)
    dfSummary = dfSummary.astype({"Opening":"string","Number of Games":int,"Win rate(%)":float,"Loss rate(%)":float,"Draw rate(%)":float, "Average Score":float,"Rating Difference": float})
    dfSummary.sort_values(by = ["Number of Games"],inplace = True,ascending = False,ignore_index=True)
    return dfSummary
        
```

```python
def PlotOpeningWinRates(dfSummary,noOpenings = 10): #Any 0% are an issue then it clips, also want to move y label further left (try 1 move)
    
    df_to_plot = dfSummary.sort_values(by = ["Number of Games"],ascending = True,ignore_index = True)
    df_to_plot = df_to_plot.tail(noOpenings).reset_index()
    df_to_plot.plot(kind = "barh",
                           stacked = True, 
                           x = "Opening", 
                           y = ["Win rate(%)","Draw rate(%)","Loss rate(%)"],
                           edgecolor = "black", 
                           color = ["white", "grey", "black"],
                           xlabel = "Percentage (%)",
                           xlim = (0,100))
    plt.title(str(f"Proportion of result for the {noOpenings} most popular openings"))
    plt.legend(loc = "best", bbox_to_anchor = (1.37,1.03))
    plt.ylabel("Openings",labelpad = 20)
    
    for row in df_to_plot.index: 
        plt.annotate(text = str(df_to_plot.loc[row,"Win rate(%)"].round(2))+"%", xy = (df_to_plot.loc[row,"Win rate(%)"]/2,row), xytext = (-1.5,-0.375), textcoords = "offset fontsize",color = "black",fontsize =  9)
        plt.annotate(text = str(df_to_plot.loc[row,"Draw rate(%)"].round(2))+"%", xy = (df_to_plot.loc[row,"Win rate(%)"]+df_to_plot.loc[row,"Draw rate(%)"]/2,row), xytext = (-1.5,-0.375), textcoords = "offset fontsize",color = "white",fontsize =  9)
        plt.annotate(text = str(df_to_plot.loc[row,"Loss rate(%)"].round(2))+"%", xy = (df_to_plot.loc[row,"Win rate(%)"]+df_to_plot.loc[row,"Draw rate(%)"]+df_to_plot.loc[row,"Loss rate(%)"]/2,row), xytext = (-1.5,-0.375), textcoords = "offset fontsize",color = "white",fontsize =  9)
        plt.annotate(text = str(df_to_plot.loc[row,"Number of Games"]) + " Games", xy = (0,row),xytext = (-1,-2), textcoords = "offset fontsize",fontsize =  7, ha = "right")
    
    plt.show()
    return
    
    
```
